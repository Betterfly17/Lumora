```python
import os, glob
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit

df = pd.read_csv("ham/HAM10000_metadata.csv")

paths = {}
for p in glob.glob("ham/**/*.jpg", recursive=True):
    paths[os.path.splitext(os.path.basename(p))[0]] = p
df["path"] = df["image_id"].map(paths)
print("missing images:", df["path"].isna().sum())

classes = sorted(df["dx"].unique())
df["label"] = df["dx"].map({c: i for i, c in enumerate(classes)})

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, val_idx = next(gss.split(df, groups=df["lesion_id"]))
train_df = df.iloc[train_idx].reset_index(drop=True)
val_df = df.iloc[val_idx].reset_index(drop=True)

overlap = set(train_df["lesion_id"]) & set(val_df["lesion_id"])
print(len(train_df), "train,", len(val_df), "test")
print("lesions in both sets:", len(overlap))
```

```python
import torch, torch.nn as nn, numpy as np
from PIL import Image
from torchvision import transforms, models
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import recall_score

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using:", device)

norm = transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
train_tf = transforms.Compose([transforms.Resize((224, 224)), transforms.RandomHorizontalFlip(),
                               transforms.RandomVerticalFlip(), transforms.ToTensor(), norm])
val_tf = transforms.Compose([transforms.Resize((224, 224)), transforms.ToTensor(), norm])

class SkinData(Dataset):
    def __init__(self, d, tf): self.d, self.tf = d, tf
    def __len__(self): return len(self.d)
    def __getitem__(self, i):
        r = self.d.iloc[i]
        return self.tf(Image.open(r["path"]).convert("RGB")), int(r["label"])

train_loader = DataLoader(SkinData(train_df, train_tf), batch_size=32, shuffle=True, num_workers=2)
val_loader = DataLoader(SkinData(val_df, val_tf), batch_size=32, num_workers=2)

model = models.mobilenet_v3_small(weights="DEFAULT")
model.classifier[3] = nn.Linear(model.classifier[3].in_features, 7)
model = model.to(device)

counts = train_df["label"].value_counts().sort_index().values
weights = torch.tensor(counts.sum() / (7 * counts), dtype=torch.float).to(device)
loss_fn = nn.CrossEntropyLoss(weight=weights)
opt = torch.optim.Adam(model.parameters(), lr=1e-4)

mel = classes.index("mel")
best = 0
for epoch in range(8):
    model.train()
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        opt.zero_grad()
        loss_fn(model(x), y).backward()
        opt.step()

    model.eval()
    preds, truth = [], []
    with torch.no_grad():
        for x, y in val_loader:
            preds += model(x.to(device)).argmax(1).cpu().tolist()
            truth += y.tolist()
    preds, truth = np.array(preds), np.array(truth)
    acc = (preds == truth).mean()
    recalls = recall_score(truth, preds, average=None, labels=range(7), zero_division=0)
    print(f"Epoch {epoch+1}: accuracy {acc:.1%} | melanoma sensitivity {recalls[mel]:.1%} | balanced {recalls.mean():.1%}")
    if recalls.mean() > best:
        best = recalls.mean()
        torch.save(model.state_dict(), "best_skin_model.pt")
        print("   saved best model")
```
